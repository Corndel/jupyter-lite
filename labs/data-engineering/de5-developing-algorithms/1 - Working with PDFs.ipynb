## Overview
In this tutorial, we’ll be working through a practical example to help you understand how to develop algorithms that can extract structured data from unstructured sources. Specifically, we’ll be focusing on extracting data from PDF files, which, in this example, contain invoice information.

Our main goal is to extract key details, such as the Order ID, Shipping Mode, and other relevant fields from these invoices and then store them in a structured format within a database.

The notebooks are in order: 1 - Working with PDFs, 2 - Pattern Matching using Regex, 3 - Full pipeline.

The sample includes a few invoices that the patterns do not match, as the full set does.
```python
# pdfplumber 0.9.0 runs in the browser; newer versions need pypdfium2, which does not
%pip install -q pdfplumber==0.9.0

import os
import pdfplumber
```

```python
# Directory containing all PDF files
pdf_directory = "data/"
pdf_files = [os.path.join(pdf_directory, file) for file in os.listdir(pdf_directory) if file.endswith(".pdf")]
```

```python
test_pdf = pdfplumber.open(pdf_files[0])
```

```python
page = test_pdf.pages[0]
```

```python
text = page.extract_text()
text
```

```python
text.split("\n")
```

```python
for line in text.split('\n'):
    if 'Order ID' in line:
        orderid = line[11:].strip()
        print(orderid)
    if 'Total' in line:
        total = line[6:].strip()
        print(total)
```

```python
order_data = []
for pdf in pdf_files:
    with pdfplumber.open(pdf) as pdf:
        for page in pdf.pages:
            text = page.extract_text()                                  # extract text
            line = text.split('\n')                                     # split the lines
            for line in text.split('\n'):                               # for each line
                if 'Order ID' in line:                                  # if the line contains 'Order ID'
                    orderid = line[11:].strip()                         # extract relavant detail
                if 'Total' in line:
                    total = line[6:].strip()
            order_data.append((orderid, total))
```
