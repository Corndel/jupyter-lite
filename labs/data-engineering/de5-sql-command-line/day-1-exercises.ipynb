## How the exercises work

Each exercise has a `%%sql` cell that starts with the exercise's instructions as SQL comments. Replace the placeholder line under the comments with your query and run the cell. The first line, `%%sql result_1a <<` (with that exercise's id), saves your query's result as a variable instead of displaying it.

Then run the check cell below it. In the original repository, `./test.sh` ran each exercise with the `sqlite3` command-line tool and compared the output with the expected output in `tests/<id>.txt`. The check cell makes the same comparison: it prints **Pass ✅** or **Fail ❌** (with the first row that differs), then shows your result. Results must match row for row, so the order of rows and columns matters.
