```python
import sqlite3
from sqlalchemy import create_engine
from sqlalchemy.pool import StaticPool

# Build the Topsify database in memory from db/reset.sql, as the repository's tests did
con = sqlite3.connect(':memory:', check_same_thread=False)
con.executescript(open('db/reset.sql', encoding='utf-8').read())

# Connect jupysql to that one connection
engine = create_engine('sqlite://', creator=lambda: con, poolclass=StaticPool)
%sql engine

# The checks that replace the repository's bash tests (see checks.py)
from checks import check
```

## How the exercises work

Each exercise has a `%%sql` cell that starts with the exercise's instructions as SQL comments. Replace the placeholder line under the comments with your SQL and run the cell. These exercises change the database, so run each one **once**; if something goes wrong, run the setup cells again to start from the original data, then redo the exercises in order.

Then run the check cell below it. In the original repository, `./test.sh` ran each exercise against a fresh database, then ran a query and compared its output with `tests/<id>.txt`. The check cell runs the same query against your database (on a copy, so it changes nothing) and prints **Pass ✅** or **Fail ❌**.
