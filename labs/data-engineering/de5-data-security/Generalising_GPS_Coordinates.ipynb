# Plotting GPS Coordinates on a Plotly Map

## ⚠️ Important: Limitations of GPS Anonymisation

This notebook shows simple techniques for anonymising location data—such as rounding coordinates or adding random noise. These are **educational examples only** and are **not guaranteed to prevent re-identification**.

They should **not** be applied to real datasets unless a proper risk assessment has been carried out.

---

### Why anonymising GPS data is especially risky

- **Population density matters**: Generalising coordinates might hide someone’s identity in a busy city but reveal it in a quieter area  
- **Patterns can still emerge**: Even ‘fuzzy’ locations can show daily routines, home and work areas, or habits  
- **Time makes it easier**: Linking location with timestamps increases the chance of identifying someone  
- **Other data can join the dots**: Public maps or leaked datasets can make re-identification easier than expected

---

### If you’re working with real location data:

- ✅ **Speak to your organisation’s Data Protection Officer (DPO) or Information Governance team**  
- ✅ **Think about who will see the data and why**, internal use and public release require different levels of protection  
- ✅ **Consider safer options**, like generating synthetic data with realistic structure and geography

---

**Always assume location data carries risk, even if names or IDs are removed.**  
If in doubt, don’t just rely on code—seek internal advice.


This notebook will allow you to visualise location coordinates on an interactive map. After importing the Plotly library, you will be prompted to enter your longitude and latitude values. Run the subsequent cells to plot on the map.

> 📗 Notes on this browser version: the maps use Plotly's `Scattermap` (the current name for the `Scattermapbox` chart type, which recent Plotly versions have removed), and the perturbation step converts metres to miles with plain arithmetic, where the original used the `geopy` library's `geodesic(meters=...).miles`. `geopy` cannot be imported in the browser. The results are the same.

```python
%pip install -q plotly==7.1.0 nbformat

import plotly.graph_objects as go
import random
```

Enter your GPS coordinates here. A quick way to find some are to google "london lat lon", for instance.

```python
# Prompt for GPS Coordinates
latitude = float(input("Enter latitude: "))
longitude = float(input("Enter longitude: "))
```

Plot the GPS coordinates

```python
# Create a Plotly figure for the original data
fig_original = go.Figure(go.Scattermap(
    lat=[latitude],
    lon=[longitude],
    mode='markers',
    marker=dict(
        size=10,
        color='blue'
    ),
    name='Original Data'
))

# Set the map style
fig_original.update_layout(
    map=dict(
        style='carto-positron',
        zoom=16,
        center=dict(lat=latitude, lon=longitude)
    )
)
# Show the map
fig_original.show()
```

Rounded the coordinate values to a set number of decimal places

### ⚠️ Coordinate Rounding Limitations

**Rounding GPS coordinates reduces precision but has important limitations:**

- **Grid-based identification**: Rounded coordinates create a grid system that may still allow identification in less populated areas
- **Precision vs. privacy trade-off**: Fewer decimal places increase privacy but reduce data utility for analysis
- **Context dependency**: The same rounding level may be appropriate for urban areas but insufficient for rural locations
- **Clustering effects**: Multiple data points in the same grid cell can reveal patterns about locations and movements

**In practice**: Always consider the population density and geographic context when choosing the appropriate level of rounding.

```python
# Specify the decimal places for rounding
decimal_places = int(input("Enter the number of decimal places: "))
# Round the coordinates to the specified decimal places
latitude_rounded = round(latitude, decimal_places)
longitude_rounded = round(longitude, decimal_places)
print(f'latitude rounded: {latitude_rounded} longitude rounded: {longitude_rounded}')
```

Plot the GPS rounded coordinates

```python
# Create a Plotly figure for the rounded data
fig_rounded = go.Figure(go.Scattermap(
    lat=[latitude_rounded],
    lon=[longitude_rounded],
    mode='markers',
    marker=dict(
        size=10,
        color='green'
    ),
    name='Rounded Data'
))
# Set the map style
fig_rounded.update_layout(
    map=dict(
        style='carto-positron',
        zoom=16,
        center=dict(lat=latitude_rounded, lon=longitude_rounded)
    )
)
# Show the map
fig_rounded.show()
```

Specify the amount to perturb the coordinate values in meters from the original location

```python
# Specify the perturbation distance in meters
perturbation_distance = float(input("Enter the perturbation distance in meters: "))
```

```python
# Convert the perturbation distance to degrees
# Metres to miles (1 mile = 1,609.344 metres), then miles to degrees (about 69 miles per degree)
perturbation_degrees = perturbation_distance / 1609.344 / 69

# Apply perturbation to the coordinates
latitude_perturbed = latitude + random.uniform(-perturbation_degrees, perturbation_degrees)
longitude_perturbed = longitude + random.uniform(-perturbation_degrees, perturbation_degrees)

print(f'latitude perturbed: {latitude_perturbed} longitude perturbed: {longitude_perturbed}')
```

```python
# Create a Plotly figure for the perturbed data
fig_perturbed = go.Figure(go.Scattermap(
    lat=[latitude_perturbed],
    lon=[longitude_perturbed],
    mode='markers',
    marker=dict(
        size=10,
        color='red'
    ),
    name='Perturbed Data'
))
# Set the map style
fig_perturbed.update_layout(
    map=dict(
        style='carto-positron',
        zoom=16,
        center=dict(lat=latitude_perturbed, lon=longitude_perturbed)
    )
)
# Show the map
fig_perturbed.show()
```
