```python
import pandas as pd # Import the pandas library for data manipulation
import statsmodels.api as sm # Import statsmodels for statistical modeling
from sklearn.model_selection import train_test_split # Import train_test_split for splitting data

ames = pd.read_csv('AmesHousing.csv') # Load the AmesHousing dataset into a pandas DataFrame
ames.head() # Display the first few rows of the DataFrame
```

```python
# Calculate the correlation matrix of the 'ames' DataFrame.
# Save to a .csv
ames.corr().to_csv('corrmat.csv')
```

```python
from sklearn.model_selection import train_test_split

# Split the ames dataset into training and testing sets.
train_data, test_data = train_test_split(ames,
                                        random_state = 42)
```

```python
# Extract the 'SalePrice' column from the train_data DataFrame.
train_dependent = train_data['SalePrice']
```

```python
import statsmodels.api as sm # Import the statsmodels library

train_independent = train_data[['Overall Qual','Lot Area']] # Create independent variable dataframe
train_independent = sm.add_constant(train_independent) # Add a constant term to the independent variables
```

```python
# Assign the 'SalePrice' column to the test_dependent variable.
test_dependent = test_data['SalePrice']
# Assign the 'Overall Qual' and 'Lot Area' columns to the test_independent variable.
test_independent = test_data[['Overall Qual','Lot Area']] 
# Add a constant to the test_independent variable.
test_independent = sm.add_constant(test_independent)
```

```python
# Import the necessary library (assuming sm is statsmodels)
import statsmodels.api as sm

# Create and fit the OLS model
model = sm.OLS(
    train_dependent,  # Dependent variable data
    train_independent # Independent variable data
).fit() # Fit the model to the data
```

```python
# Print the model summary.
model.summary()
```

# Model Evaluation

```python
# Apply the model to make predictions on the test data.
test_data['Predicted SalePrice'] = model.predict(test_independent)
```

```python
from sklearn import metrics # Import the 'metrics' module from the 'sklearn' library.
```

```python
# Calculate Mean Absolute Error (MAE)
mae = metrics.mean_absolute_error(test_data['SalePrice'], 
                                  test_data['Predicted SalePrice'])

print(round(mae))
```

```python
# Calculate the Mean Squared Error (MSE) between 'SalePrice' and 'Predicted SalePrice' columns in the test_data DataFrame.
mse = metrics.mean_squared_error(test_data['SalePrice'], test_data['Predicted SalePrice'])


print(round(mse))
```

```python
import numpy as np

np.sqrt(mse).round() # Calculate the square root of the mean squared error (mse).
```
