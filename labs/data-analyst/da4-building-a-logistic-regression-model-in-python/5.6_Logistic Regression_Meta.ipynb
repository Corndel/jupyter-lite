# 5.6 When is a logistic Regression appropriate

```python
meta_stock.columns
```

```python
meta_stock[['Open', 
            'Close',
            'Change in stock price',
            'Actual up/down tomorrow'
           ]].groupby('Actual up/down tomorrow').mean()
```

```python
import scipy.stats as stats

group1 = meta_stock[meta_stock['Actual up/down tomorrow'] == 0]

group2 = meta_stock[meta_stock['Actual up/down tomorrow'] == 1]

stats.ttest_ind(group1['Change in stock price'], group2['Change in stock price'])
```
