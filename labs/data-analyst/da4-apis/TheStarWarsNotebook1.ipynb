# Importing the required libraries

In order to be able to interrogate and present information from the API into this client, we are going to load in a couple of libraries. Libraries are pre-written and contain bundles of code which can aid the functionality of the program. Here, we will load in the JSON ( JavaScript Object Notation ) library to be able to return the data in a JSON format and the Request library will will allow us to make HTTP requests to find the information

```python
import json
import requests
```

# Checking the connection

The following code will make a connection to the API and store it as a variable called response. We can then call this variable to find out information such as whether the request has been successful by printing the status code. A status code of 200 means that the connection has been successfully made. 

```python
# We will set up a variable called r (short for response, so that we don't have to keep typing response!!)
r = requests.get('https://swapi.dev/api/people/')
```

We can run the following code to ensure that the connection has been made successfully. The 200 response code will indicate that the connection has been made. There is a list of the status codes in Explore further section of the API Rise module that will explain more about this. We can use a PRINT function to output the response status code to screen.

```python
print(r)
```

```python
print(r.json())
```

The code in the cell below will print out the results of the API in JSON format. JSON outputs the results as key pair values. If you look carefully at the cell above, you can see, reading from the left that there are two values joined by a semi-colon. The pairs are as follows:

* 'count': 82
* 'next': 'https://swapi.dev/api/people/?page=2'
* 'previous': None
* 'results':[ a very large list]

We are now going to print out the contents of the list ( which is called results ) to get just the information on the characters. As you can see there is a list of characters with their attibutes. If we look at Luke Skywalker, we can see that as well as his physical attributes, there are links to other tables where other information is held such as home planets etc. 



```python
print(r.json()['results'])
```

The code below will iterate over the json object and print out the names of the characters in the film. As you can see there have been 10 character's names output. Obviously there are a lot more characters in this film but the API has only returned 10 due to pagination, where the very large data set has been split into smaller chunks. 

```python
for data in r.json()['results']:
    print(data['name'])
```

# Writing a function to output the characters

```python
def crawl():                                              # define a function
    response=requests.get('https://swapi.dev/api/people') # get the response
    api_results=json.loads(response.content)              # create a variable for the API results
    return (api_results)                                  # return the API results
```
