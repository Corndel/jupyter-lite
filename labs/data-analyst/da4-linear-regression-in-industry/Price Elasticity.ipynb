# Calculating price elasticity of milk across two markets

A supermarket chain sells across two markets: Europe and Asia. They are considering an increase in the price of milk, to account for the increasing costs of the supply chains. Before they increase the prices, the economists advising the supermarket CEO want to study the price elasticity of milk across the two markets, so see which market will be more heavily impacted by the increase.

```python
# Start by reading the data from PriceElasticity.xlsx

import pandas as pd

milk_price_demand = pd.read_excel('PriceElasticity.xlsx')
milk_price_demand.head()
```

```python
# Discover the correlations between the fields in the data

milk_price_demand.corr()
```

There is a strong correlation between the Price per litre and the Sales across each region (measured in millions of litres). We can therefore use the price to build a linear regression for sales in each region.

This will mean building two separate linear regression models: one for Europe and one for Asia.

##### Building the Europe model

```python
dependent = milk_price_demand['Europe Sales (Ml)']
independent = milk_price_demand['Price / l']
```

```python
import statsmodels.api as sm

independent = sm.add_constant(independent)
```

```python
import statsmodels.api as sm

# Build the model

model_eu = sm.OLS(
    dependent, 
    independent
).fit()

# Print model summary statistics

model_eu.summary()
```

##### Building the Asia model

```python
dependent = milk_price_demand['Asia Sales (Ml)']
independent = milk_price_demand['Price / l']
```

```python
import statsmodels.api as sm

independent = sm.add_constant(independent)
```

```python
import statsmodels.api as sm

# Build the model

model_as = sm.OLS(
    dependent, 
    independent
).fit()

# Print model summary statistics

model_as.summary()
```

##### Calculating Price Elasticity of Demand

The Price Elasticity of Demand (PED) is calculated as:

$$\text{PED} = \frac{\text{current price}}{\text{current quantity}}\times \text{Coefficient of Price}$$

The current average price of milk in Europe and Asia is €1.43. We can calculate the price elasticity in each market:

**Europe:**
$$\text{PED} = \frac{1.43}{10210 - 5970 \times 1.43} \times (-5970)$$

```python
1.43/(10210-5970*1.43)*(-5970)
```

**Asia:**
$$\text{PED} = \frac{1.43}{7139 - 2893 \times 1.43} \times (-2893)$$

```python
1.43/(7139-2893*1.43)*(-2893)
```
